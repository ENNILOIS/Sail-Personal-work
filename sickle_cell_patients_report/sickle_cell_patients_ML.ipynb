```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

print("Libraries loaded successfully!")
```

```python
df = pd.read_csv("sickle_cell_patients_report.csv")

print("Data loaded successfully!")
print("Number of students:", len(df))

df.head(7)
```

```python
print("Columns in our dataset:")
print(df.columns.tolist())

print()

print("Firt 5 patient records:")
print(df.head())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Number of sickle cell patients:", (df["Sickle_Cell_Status"] == 1).sum())
print("Number of non-sickle cell patients:", (df["Sickle_Cell_Status"] == 0).sum())
```

```python
X = df[["Age_Months", "Father_SCT", "Mother_SCT", "Family_History_SCD", "Fatigue", "Jaundice", "Swelling_Hands_Feet", "Frequent_Infections", "Pain_Crises_Ever", "Hemoglobin_g_dL", "Hb_Electrophoresis_HbS_Percentage"]]

y = df["Sickle_Cell_Status"]

print("Our input features (X):")
print(X.head())

print("Our target (y):")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.05,
    random_state=42
)

print("Training patient records:", len(X_train))
print("Testing patient records:", len(X_test))
```

```python
model = DecisionTreeClassifier(
    max_depth=10,
    random_state=42
)

model.fit(X_train, y_train)

print("Model training complete!")
```

```python
y_pred = model.predict(X_test)

print("Predictions made successfully!")
print("Number of predictions:", len(y_pred))

print()
print("Actual answers:")
print(y_test.values)

print()
print("Model predictions:")
print(y_pred)
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()

print("True Negative (tn)  :", tn)
print("False Positive (fp) :", fp)
print("False Negative (fn) :", fn)
print("True Positive (tp)  :", tp)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 1), "%")
```

```python
precision = precision_score(y_test, y_pred)

print("Precision:", round(precision * 100, 1), "%")
```

```python
recall = recall_score(y_test, y_pred)

print("Recall:", round(recall * 100, 1), "%")
```

```python
f1 = f1_score(y_test, y_pred)

print("F1:", round(f1 * 100, 1), "%")
```

```python
print("====== MODEL EVALUATION ======")

print("Accuracy  :", round(accuracy * 100, 1), "%")
print("Precision :", round(precision * 100, 1), "%")
print("Recall    :", round(recall * 100, 1), "%")
print("F1        :", round(f1 * 100, 1), "%")

print()
print("Confusion Matrix:")
print(cm)
```
