#### VISUALIZATION

##### QUESTION 1: Do Both Patrents' SCT Statuses relate to SCD Status?

```python
parent_status = df.groupby(
    ["Mother_SCT", "Father_SCT"]
)["Sickle_Cell_Status"].value_counts().unstack(fill_value = 0)

parent_status = parent_status.rename(
    columns= {0: "No SCD", 1: "Yes SCD"}
)

fig, ax = plt.subplots(figsize = (13, 6))
fig.patch.set_facecolor("#e6f2ff")
ax.set_facecolor("#e6f2ff")

parent_status.plot(kind = "bar",
                    color = ["black", "skyblue"],
                    width = 0.8,
                    ax=ax)

for container in ax.containers:
    ax.bar_label(container, fontsize = 11, fontweight = "bold", color = "blue")

ax.set_title("MOTHER & FATHER SCT STATUS VS SCD STATUS", 
             fontsize = 14,
             fontweight = "bold",
             fontstyle = "italic",
             color = "blue"
             )

ax.set_xlabel("Parents SCT Status",
               fontsize = 13,
               color = "blue",
               fontstyle = "italic")

ax.set_ylabel("Number of Parents", 
              fontsize = 13, 
              color = "blue",
              fontstyle = "italic")

ax.tick_params(axis = "x", rotation = 35)

plt.tight_layout()
plt.show()
```

##### QUESTION 2: What is the Most Common SCD Complications?

```python
df_clean.columns
```

```python
complications = [
    "Fatigue",
    "Jaundice",
    "Swelling_Hands_Feet",
    "Frequent_Infections",
    "Pain_Crises_Ever"
]

complication_counts = {}

for column in complications:
    complication_counts[column] = df[column].value_counts()[1]

complication_counts = pd.Series(complication_counts).sort_values(ascending=False)

fig,ax = plt.subplots(figsize = (14, 5))

sns.barplot(
    x = complication_counts.index,
    y = complication_counts.values,
    palette = "muted",
    legend = False,
    ax=ax
)

ax.set_title("MOST COMMON SCD COMPLICATIONS",
              fontsize = 13,
              fontweight = "bold")

ax.set_xlabel("COMPLICATION",
               fontsize = 11,
               fontstyle = "italic")

ax.set_ylabel("NUMBERS OF PATIENTS",
                fontsize = 11,
                fontstyle = "italic")

ax.tick_params(axis = "x", rotation = 25)

plt.tight_layout()
plt.show()
```
